# Model Training

```python
from ultralytics import YOLO

fine_tuned_model = YOLO('yolo26x.pt')

fine_tuned_model.train(
    data='data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    freeze=10,
    lr0=0.001,
    lrf=0.01,
    warmup_epochs=3,
    mixup=1,
    cutmix=1,
    project='bottle_detection',
    name='v1_specialized'
)
```

```python
fine_tuned_model_v2 = YOLO('yolo26s.pt')

fine_tuned_model_v2.train(
    data='data.yaml',
    epochs=100,
    patience=20,
    imgsz=640,
    batch=32,
    freeze=10,
    lr0=0.001,
    lrf=0.01,
    warmup_epochs=3,
    mosaic=0.5,
    mixup=0.1,
    cutmix=0.1,
    project='bottle_detection',
    name='v2_specialized'
)
```

```python
import shutil
from google.colab import files

relative_source = 'runs/detect/bottle_detection'

if os.path.exists(relative_source):
    shutil.make_archive('all_bottle_runs_v3', 'zip', relative_source)
    files.download('all_bottle_runs_v3.zip')
    print("Download triggered successfully.")
else:
    print(f"Error: Could not find {relative_source}. Check your folder structure!")
```
